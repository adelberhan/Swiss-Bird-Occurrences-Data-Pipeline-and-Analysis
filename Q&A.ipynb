```python
import pandas as pd
```

```python
df =pd.read_csv('data/csv/swiss_bird_occurrences.csv')
```

```python
top_20_species = df['species_name'].value_counts().head(20)

print("Top 20 species with the most occurrences:")
print(top_20_species)
```

```python
occurrences_by_year = df["observation_year"].value_counts().sort_index()


print("Occurrences by year:")
print(occurrences_by_year)
```

```python
# occurrences per year, 
# how many records have no species identification.

occurrences_per_year = df["species_name"].isna().sum()

print(f"Number of records with no species identification: {occurrences_per_year}")
```
